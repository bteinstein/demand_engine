```python
import duckdb 
H3_DUCKDB_PATH = STORAGE_CONFIG['h3_duckdb_path']


def get_duckdb_schemas(db_path: str) -> dict:
    """
    Extracts the schema (columns + types) of all tables in a DuckDB database.

    Args:
        db_path (str): Path to the DuckDB file.

    Returns:
        dict: {table_name: {column_name: data_type, ...}, ...}
    """
    conn = duckdb.connect(db_path)
    
    # Get list of all tables in the database
    tables = conn.execute(
        "SELECT table_schema, table_name FROM information_schema.tables WHERE table_type='BASE TABLE';"
    ).fetchall()
    
    schema_dict = {}
    for schema, table in tables:
        # Get schema info for each table
        cols = conn.execute(f"PRAGMA table_info('{schema}.{table}')").fetchall()
        # Structure: [(cid, name, type, notnull, dflt_value, pk), ...]
        schema_dict[f"{schema}.{table}"] = {
            col[1]: col[2] for col in cols
        }
    
    conn.close()
    return schema_dict

get_duckdb_schemas(db_path = H3_DUCKDB_PATH)
```

```python
import pandas as pd
import geopandas as gpd
import h3
import numpy as np
from typing import Dict, Any, Tuple
from shapely.strtree import STRtree

def assign_customers_to_clusters(grid_data: Dict[str, Any], customers_gdf: gpd.GeoDataFrame) -> pd.DataFrame:
    """
    Assign customers to H3 clusters using 3-tier approach
    
    Args:
        grid_data: Dict containing 'h3_resolution', 'clipped_cells', 'cell_geometries'
        customers_gdf: GeoDataFrame with customer_id and geometry columns
    
    Returns:
        DataFrame with customer assignments and confidence levels
    """
    return_cols = ['customer_id', 'cluster_id', 'h3_cell_id', 'assignment_confidence', 'assignment_tier']
    assignment_stats = {'tier1': 0, 'tier2': 0, 'tier3': 0}
    
    if customers_gdf.empty:
        return pd.DataFrame(columns=return_cols)
    
    resolution = grid_data['h3_resolution']
    clipped_cells = set(grid_data['clipped_cells'])
    cell_geometries = grid_data['cell_geometries']
    
    # Extract coordinates and IDs
    coords = np.column_stack([customers_gdf.geometry.y, customers_gdf.geometry.x])
    customer_ids = customers_gdf['customer_id'].values
    
    # Tier 1: H3 cell exact match
    tier1_assignments = []
    unassigned_mask = np.ones(len(customers_gdf), dtype=bool)
    
    for idx, (lat, lng) in enumerate(coords):
        try:
            customer_h3_cell = h3.latlng_to_cell(lat, lng, resolution)
            
            # Only assign if customer's H3 cell exists in clipped_cells
            if customer_h3_cell in clipped_cells:
                tier1_assignments.append({
                    'customer_id': customer_ids[idx],
                    'cluster_id': customer_h3_cell,
                    'h3_cell_id': customer_h3_cell,
                    'assignment_confidence': 1.0,
                    'assignment_tier': 'h3_inclusion'
                })
                unassigned_mask[idx] = False
                assignment_stats['tier1'] += 1
        except Exception:
            continue
    
    # Tier 2: Point-in-polygon with spatial index
    tier2_assignments = []
    if cell_geometries and np.any(unassigned_mask):
        geometries = list(cell_geometries.values())
        cell_ids = list(cell_geometries.keys())
        spatial_index = STRtree(geometries)
        
        unassigned_customers = customers_gdf[unassigned_mask]
        
        for idx, (_, customer) in enumerate(unassigned_customers.iterrows()):
            original_idx = np.where(unassigned_mask)[0][idx]
            customer_point = customer.geometry
            
            # Query spatial index
            possible_matches_idx = list(spatial_index.query(customer_point))
            
            for geom_idx in possible_matches_idx:
                if geometries[geom_idx].contains(customer_point):
                    cell_id = cell_ids[geom_idx]
                    tier2_assignments.append({
                        'customer_id': customer_ids[original_idx],
                        'cluster_id': cell_id,
                        'h3_cell_id': cell_id,
                        'assignment_confidence': 0.8,
                        'assignment_tier': 'point_in_polygon'
                    })
                    unassigned_mask[original_idx] = False
                    assignment_stats['tier2'] += 1
                    break
    
    # Tier 3: Manual review for unassigned
    tier3_assignments = []
    remaining_indices = np.where(unassigned_mask)[0]
    
    for idx in remaining_indices:
        tier3_assignments.append({
            'customer_id': customer_ids[idx],
            'cluster_id': None,
            'h3_cell_id': None,
            'assignment_confidence': 0.0,
            'assignment_tier': 'manual_review'
        })
        assignment_stats['tier3'] += 1
    
    # Combine all assignments
    all_assignments = tier1_assignments + tier2_assignments + tier3_assignments
    result_df = pd.DataFrame(all_assignments)
    
    # Print stats
    total = sum(assignment_stats.values())
    if total > 0:
        print(f"Assignment stats - Tier1: {assignment_stats['tier1']/total:.1%}, "
              f"Tier2: {assignment_stats['tier2']/total:.1%}, "
              f"Tier3: {assignment_stats['tier3']/total:.1%}")
    
    return result_df

#  usage:
filtered_grid_data = { key:value for key, value in grid_results_.items() if key ==  '1647109'}
# """
# # Sample grid_data structure
# grid_data = {
#     'h3_resolution': 8,
#     'clipped_cells': ['8c2a1072b1b7fff', '8c2a1072b1b3fff', ...],  # list of h3 cell IDs
#     'cell_geometries': {
#         '8c2a1072b1b7fff': polygon_geometry_1,
#         '8c2a1072b1b3fff': polygon_geometry_2,
#         ...
#     }
# }

# # Sample customers_gdf
# customers_gdf = gpd.GeoDataFrame({
#     'customer_id': [1, 2, 3, ...],
#     'geometry': [Point(lng1, lat1), Point(lng2, lat2), ...]
# })

# # Run assignment
# assignments = assign_customers_to_clusters(grid_data, customers_gdf)
# """
```

```python
# 4. Export for deployment: ETA: 1 Min
PILOT_SPS_CLUSTER_R8_FLAT = clusterer.export_results(PILOT_SPS_CLUSTER_R8, output_format="csv") 


def extract_coverage_and_assignment_results(clusterer, cluster_result_dict):
    PILOT_SPS_CLUSTER_R8_FLAT = clusterer.export_results(cluster_result_dict, output_format="csv")

    # Stock Point Assignment Summary
    df_output_sp_coverage_cluster = PILOT_SPS_CLUSTER_R8_FLAT['territory_cells']  
    df_output_sp_customer_assignment = PILOT_SPS_CLUSTER_R8_FLAT['assignments']
    df_output_ap_coverage_cluster_summary = PILOT_SPS_CLUSTER_R8_FLAT['clusters']

    df_output_sp_customer_assignment['stock_point_id'] = df_output_sp_customer_assignment['stock_point_id'].astype(int)
    df_output_sp_coverage_cluster['stock_point_id'] = df_output_sp_coverage_cluster['stock_point_id'].astype(int)

    return df_output_sp_coverage_cluster, df_output_sp_customer_assignment, df_output_ap_coverage_cluster_summary

def prepare_sp_assignment_summary(df_output_sp_customer_assignment, sp_dim_df):
    sp_assignment_summary = (df_output_sp_customer_assignment
                            .groupby(['stock_point_id','cluster_id'])['customer_id'].count()
                            .reset_index(name='n_customers')
                            .rename({'cluster_id':'h3_cell'}, axis=1) 
                        ) 
    # Stock Point Coverage - Assignment Summary
    sp_coverage_cluster_and_assignment_summary = (df_output_sp_coverage_cluster
                                                .merge(sp_dim_df[['stock_point_id', 'stock_point_name']] , on='stock_point_id', how='left')
                                                .merge(sp_assignment_summary, how='left', on=['stock_point_id','h3_cell'])
                                                .fillna({'n_customers':0})
                                                .rename({'h3_cell':'cluster_id'}, axis=1)
                                                ) 
    sp_coverage_cluster_and_assignment_summary['n_customers'] = sp_coverage_cluster_and_assignment_summary['n_customers'].astype(int) 

    return sp_assignment_summary


df_output_sp_coverage_cluster, df_output_sp_customer_assignment, df_output_sp_coverage_cluster_summary = extract_coverage_and_assignment_results(clusterer = clusterer, 
                                                                                                                                                 cluster_result_dict = PILOT_SPS_CLUSTER_R8)
sp_coverage_cluster_and_assignment_summary = prepare_sp_assignment_summary(df_output_sp_customer_assignment, sp_dim_df)

 
print(df_output_sp_coverage_cluster.stock_point_id.nunique())
print(sp_coverage_cluster_and_assignment_summary.stock_point_id.nunique())
```

```python
import folium
import h3

# H3 cell ID
h3_cell = "8858838323fffff"

# Customer coordinates (Customer_ID, Lat, Lng)
customers = [
    (4113313, 7.3458061, 3.8757119),
    (5472323, 7.35773, 3.8721093),
    (5123649, 7.3658342, 3.8151936)
]

# Get H3 polygon boundary
boundary = h3.cell_to_boundary(h3_cell)

# Center map on H3 cell
lat, lng = h3.cell_to_latlng(h3_cell)
m = folium.Map(location=[lat, lng], zoom_start=13)

# Add H3 polygon
folium.Polygon(
    locations=boundary,
    color="blue",
    weight=2,
    fill=True,
    fill_opacity=0.1,
    popup=f"H3 Cell: {h3_cell}"
).add_to(m)

# Add customer markers
for cust_id, lat, lon in customers:
    folium.Marker(
        location=[lat, lon],
        popup=f"Customer ID: {cust_id}",
        icon=folium.Icon(color="red", icon="user")
    ).add_to(m)

# Save to HTML
m.save("h3_customers_map.html")
print("Map saved to h3_customers_map.html")
```
